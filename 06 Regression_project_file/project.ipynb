# Step 3 - Seperate X and Y


```python
X = df.drop(columns=["Id", "SalePrice"])
Y = df[["SalePrice"]]
```

```python
X.head()
```

```python
Y.head()
```

 # Step 4 - Preprocess data for Feature Selection



```python
cat = list(X.columns[X.dtypes == "object"])
con = list(X.columns[X.dtypes != "object"])
```

```python
cat
```

```python
con
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer
```

```python
num_pipe1 = make_pipeline(SimpleImputer(strategy="median"), StandardScaler())
```

```python
cat_pipe1 = make_pipeline(
    SimpleImputer(strategy="constant", fill_value="Not_Avail"), OrdinalEncoder()
)
```

```python
pre1 = ColumnTransformer([("num", num_pipe1, con), ("cat", cat_pipe1, cat)]).set_output(
    transform="pandas"
)
```

```python
X_pre = pre1.fit_transform(X)
X_pre.head()
```

# Backward Feature selection

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LinearRegression

base_model = LinearRegression()

sel = SequentialFeatureSelector(
    base_model, n_features_to_select="auto", direction="backward", n_jobs=-1
)
sel.fit(X_pre, Y)

imp_cols = sel.get_feature_names_out()
print(imp_cols)
```

```python
sel_cols = [col.split("__")[1] for col in imp_cols]
sel_cols
```

```python
X_sel = X[sel_cols]
X_sel.head()
```

# Feature selection is complete


# Step 5 - Apply final preprocessing on Selected features

Apply OneHotEncoding

```python
cat_sel = list(X_sel.columns[X_sel.dtypes == "object"])
con_sel = list(X_sel.columns[X_sel.dtypes != "object"])
```

```python
cat_sel
```

```python
con_sel
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
num_pipe2 = make_pipeline(SimpleImputer(strategy="median"), StandardScaler())
```

```python
cat_pipe2 = make_pipeline(
    SimpleImputer(strategy="constant", fill_value="Not_Avail"),
    OneHotEncoder(handle_unknown="ignore", sparse_output=False, drop="first"),
)
```

```python
pre2 = ColumnTransformer(
    [("num", num_pipe2, con_sel), ("cat", cat_pipe2, cat_sel)]
).set_output(transform="pandas")
```

```python
X_sel_pre = pre2.fit_transform(X_sel)
X_sel_pre.head()
```

# Step 6 - Train Test Split


```python
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(
    X_sel_pre, Y, test_size=0.2, random_state=20
)
```

# Setp 8 - Selected Model Evaluation

```python
from sklearn.metrics import (
    root_mean_squared_error,
    mean_absolute_error,
    mean_absolute_percentage_error,
    r2_score,
)
```

```python
def evaluate_model(model, x, y):
    ypred = model.predict(x)

    rmse = root_mean_squared_error(y, ypred)
    mae = mean_absolute_error(y, ypred)
    mape = mean_absolute_percentage_error(y, ypred)
    r2 = r2_score(y, ypred)

    print(f"RMSE : {rmse:.2f}")
    print(f"MAE : {mae:.2f}")
    print(f"MAPE : {mape:.2%}")
    print(f"R2 : {r2:.2%}")
```

```python
print("Train Results : ")
evaluate_model(best_ridge, xtrain, ytrain)
```

```python
print("Test Results Ridge : ")
evaluate_model(best_ridge, xtest, ytest)
```
